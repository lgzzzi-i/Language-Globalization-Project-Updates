Determining how many languages categorized as NaN endangerment level

```python
nan_df = language_df.iloc[language_df["endangerment"].isna()]
len(nan_df)
```

```python
palette = {
    "endangerment" : {
        'awakening': "#eece9f",
        'endangered' : "#8e6c2f",
        'severely endangered' : "#445128",
        'dormant' : "#283618",
        'vulnerable' :   "#cd8742",
        'critially endangered' : "#606c38",
        'at risk' : "#dda15e",
        'threatened' : "#bc6c25",
        'nan' : "#7a4a5c"
    },
    
    "main": {
        "black_forest": "#283618",
        "olive_leaf": "#606c38",
        "copper": "#bc6c25",
        "light_caramel": "#dda15e",
        "cornsilk": "#fefae0",
    },
    "extra": {
        "sage": "#a3b18a",
        "butter": "#efd27a",
        "rust": "#8f3b1b",
        "teal": "#3f7d78",
        "plum": "#7a4a5c",
        "stone": "#c9c7ae",
        "espresso": "#3a2a1a",
    }
}
```

```python
fig = px.scatter_geo(
    language_df,
    lat = "Latitude",
    lon = "Longitude",
    hover_name = "Name",
    color = "endangerment",
    color_discrete_map=palette["endangerment"]
)

fig.update_layout(
    title = "Language Endangerment by Region",
    legend_title = "Degree of Endangerment"
)

fig.show()
```
